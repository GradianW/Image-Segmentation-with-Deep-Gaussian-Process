## 📌 Fondasi 0: Notasi Baku & Konvensi Matematika Tesis Damianou

Sebelum kita mendalami bab-bab spesifik, berikut rangkuman konvensi notasi baku yang digunakan Andreas Damianou sepanjang tesisnya agar tidak membingungkan saat menelaah persamaan:

### 1. Simbol & Variabel Utama
| Simbol | Dimensi | Deskripsi |
|---|---|---|
| $N$ | $\mathbb{N}$ | Jumlah titik data observasi |
| $D$ | $\mathbb{N}$ | Dimensi observasi output ($Y \in \mathbb{R}^{N \times D}$) |
| $Q$ | $\mathbb{N}$ | Dimensi variabel input / ruang laten ($X \in \mathbb{R}^{N \times Q}$) |
| $M$ | $\mathbb{N}$ | Jumlah titik induksi (*inducing points*), umumnya $M \ll N$ |
| $Y = [y_1, \dots, y_N]^\top$ | $\mathbb{R}^{N \times D}$ | Matriks data observasi/target |
| $X = [x_1, \dots, x_N]^\top$ | $\mathbb{R}^{N \times Q}$ | Matriks input observasi atau variabel laten |
| $F = [f_1, \dots, f_N]^\top$ | $\mathbb{R}^{N \times D}$ | Nilai fungsi laten GP yang dievaluasi pada titik data $X$ |
| $Z = [z_1, \dots, z_M]^\top$ | $\mathbb{R}^{M \times Q}$ | Lokasi titik induksi (*inducing inputs* / pseudo-inputs) |
| $U = [u_1, \dots, u_M]^\top$ | $\mathbb{R}^{M \times D}$ | Nilai fungsi laten pada inducing points: $u_m = f(z_m)$ |
| $\beta = \sigma_y^{-2}$ | $\mathbb{R}^+$ | Presisi noise observasi Gaussian ($y_n = f(x_n) + \epsilon_n, \; \epsilon_n \sim \mathcal{N}(0, \beta^{-1} I)$) |

### 2. Matriks Kovariansi Kernel
* $K_{NN} = k(X, X) \in \mathbb{R}^{N \times N}$: Matriks kovariansi antar titik data observasi.
* $K_{MM} = k(Z, Z) \in \mathbb{R}^{M \times M}$: Matriks kovariansi antar titik induksi.
* $K_{NM} = k(X, Z) \in \mathbb{R}^{N \times M}$: Matriks kovariansi silang antara data observasi dan inducing points ($K_{MN} = K_{NM}^\top$).

### 3. Statistik $\Psi$ (*Psi-Statistics*) untuk Variational Marginalisation
Ketika input $X$ tidak pasti (*uncertain*) atau merupakan variabel acak laten dengan distribusi variasional Gaussian $q(X) = \prod_{n=1}^N \mathcal{N}(x_n \mid \mu_n, S_n)$:
$$\psi_0 = \operatorname{Tr}\left( \mathbb{E}_{q(X)}[K_{NN}] \right) = \sum_{n=1}^N \mathbb{E}_{q(x_n)}[k(x_n, x_n)]$$
$$\Psi_1 = \mathbb{E}_{q(X)}[K_{NM}] \in \mathbb{R}^{N \times M}, \quad [\Psi_1]_{n, m} = \mathbb{E}_{q(x_n)}[k(x_n, z_m)]$$
$$\Psi_2 = \mathbb{E}_{q(X)}[K_{MN} K_{NM}] = \sum_{n=1}^N \mathbb{E}_{q(x_n)}[k(z, x_n) k(x_n, z^\top)] \in \mathbb{R}^{M \times M}$$

> **Catatan Penting**: Tiga besaran $\psi_0, \Psi_1, \Psi_2$ di atas adalah inovasi analitik utama Damianou (bersama Michalis Titsias & Neil Lawrence) yang memungkinkan marginalisasi analitik (*closed-form*) untuk meloloskan distribusi probabilitas melewati fungsi kovariansi non-linear GP tanpa perlu sampling deterministik.

## 💬 Sesi 1: Perbedaan & Evolusi — GP Standar, Sparse GP, Variational Sparse GP, hingga Variational GP-LVM

### ❓ Pertanyaan Mahasiswa
> *"Saya sudah paham tentang GP biasa, yaitu regression. Sekarang saya bingung karena ada sparse, variational sparse, variational GP-LVM, dan lain-lain. Hal-hal tersebut cukup asing dan saya tidak paham apa bedanya."*

### 💡 Penjelasan & Derivasi Matematis
Untuk memahami mengapa ada begitu banyak varian (*Sparse*, *Variational Sparse*, *Variational GP-LVM*), kita cukup melihatnya sebagai **dua lompatan besar dalam memecahkan dua masalah mendasar GP standar**:

1. **Lompatan 1 (Masalah Skalabilitas Komputasi)**: Bagaimana menangani data besar ketika invers matriks kovariansi membutuhkan $\mathcal{O}(N^3)$? $\implies$ **Sparse GP & Variational Sparse GP**.
2. **Lompatan 2 (Masalah Sifat Input $X$)**: Bagaimana jika input $X$ bukan angka pasti yang diketahui, melainkan tidak pasti (*uncertain*), berisik, atau bahkan variabel laten tersembunyi yang ingin kita cari tahu? $\implies$ **Variational GP-LVM & Deep GP**.

---

### 🗺️ Ringkasan Perbandingan 4 Paradigma GP

| Metode | Status Input $X$ | Cara Kerja Inferensi | Kompleksitas Waktu | Karakteristik & Limitasi Utama |
|---|---|---|---|---|
| **1. Standard GP Regression (GPR)** | Diketahui & Pasti (Deterministik) | Eksak: $p(y_* \mid x_*, \mathcal{D}) = \mathcal{N}(\mu_*, \sigma_*^2)$ | $\mathcal{O}(N^3)$ | Akurat sempurna, tetapi macet pada $N > 5000$ karena membalik matriks $(K_{NN} + \sigma_y^2 I)_{N \times N}$. |
| **2. Sparse GP Klasik** *(SoR, DTC, FITC)* | Diketahui & Pasti (Deterministik) | Aproksimasi Kovariansi: Mengganti prior GP dengan prior termodifikasi berbasis $M$ *inducing points* $Z$. | $\mathcal{O}(NM^2)$ | Cepat ($M \ll N$), namun **rentan overfitting** jika lokasi $Z$ dioptimasi langsung via gradien marginal likelihood aproksimasi. |
| **3. Variational Sparse GP** *(Titsias 2009)* | Diketahui & Pasti (Deterministik) | Variational Inference: Mempertahankan prior GP eksak, mengoptimasi lower bound (ELBO) $\mathcal{L}_{SGPR}$. | $\mathcal{O}(NM^2)$ | **Bebas overfitting pada $Z$** (karena $Z$ adalah parameter variasional), memiliki penalti ketidakpastian jejak matriks $\operatorname{Tr}(K_{NN} - Q_{NN})$. |
| **4. Variational GP-LVM** *(Damianou Bab 3)* | **Tidak Diketahui** (Variabel Acak Laten $X \sim p(X)$) | Memperluas Titsias dengan mengintegralkan distribusi variasional $q(X)$ secara analitik via statistik $\Psi$. | $\mathcal{O}(NM^2)$ | Mampu melakukan reduksi dimensi non-linear probabilistik dan mempropagasi distribusi input $q(X)$ ke fungsi kernel. |
| **5. Deep GP** *(Damianou Bab 6)* | Berlapis-lapis: output layer $l$ menjadi input tak pasti bagi layer $l+1$. | Komposisi proses hierarkis dengan propagasi ketidakpastian antar-layer. | $\mathcal{O}(L \cdot NM^2)$ | Representasi non-linear yang sangat kaya (mirip Deep Neural Network) namun memiliki ketidakpastian Bayesian di setiap layer. |

### 🔍 Pembahasan Mendalam Masing-Masing Konsep

#### 1. Titik Tolak: Mengapa GP Regresi Biasa Tidak Cukup?
Pada GP regresi standar:
$$y_n = f(x_n) + \epsilon_n, \quad \epsilon_n \sim \mathcal{N}(0, \sigma_y^2)$$
Distribusi fungsi laten $F = [f(x_1), \dots, f(x_N)]^\top$ memiliki prior:
$$p(F \mid X) = \mathcal{N}(F \mid 0, K_{NN})$$
Log-marginal likelihood yang dimaksimalkan untuk melatih hyperparameter kernel $\theta$ adalah:
$$\log p(Y \mid X) = -\frac{1}{2} Y^\top (K_{NN} + \sigma_y^2 I_N)^{-1} Y - \frac{1}{2} \log \det(K_{NN} + \sigma_y^2 I_N) - \frac{N}{2} \log 2\pi$$

> ⚠️ **Hambatan Fatal**:
> Invers $(K_{NN} + \sigma_y^2 I_N)^{-1}$ butuh dekomposisi Cholesky berukuran $N \times N$ dengan biaya $\mathcal{O}(N^3)$ dan memori $\mathcal{O}(N^2)$. Untuk citra segmentasi ukuran $256 \times 256$ piksel ($N = 65.536$), GP biasa membutuhkan memori puluhan Gigabyte dan waktu komputasi yang tidak realistis!

---

#### 2. Sparse GP Klasik: Gagasan "Titik Induksi" (*Inducing Points*)
Untuk mengatasi $\mathcal{O}(N^3)$, para peneliti (seperti Snelson & Ghahramani 2006, Quiñonero-Candela & Rasmussen 2005) memperkenalkan **$M$ buah titik semu / titik induksi** (*inducing points*):
$$Z = [z_1, z_2, \dots, z_M]^\top \in \mathbb{R}^{M \times Q}, \quad \text{di mana } M \ll N$$
Fungsi laten pada titik induksi ini dinamakan $U = [u_1, \dots, u_M]^\top = [f(z_1), \dots, f(z_M)]^\top$.

* **Analogi**: Bayangkan $N$ adalah seluruh warga satu negara ($N = 270$ juta). Untuk memprediksi preferensi mereka, Anda tidak perlu mensurvei setiap warga, cukup pilih $M = 500$ responden representatif ($Z$) yang tersebar strategis di ruang input!
* **Aproksimasi Kovariansi Nyström**:
  $$K_{NN} \approx Q_{NN} = K_{NM} K_{MM}^{-1} K_{MN}$$
* **Kelemahan Sparse GP Klasik (FITC/DTC/SoR)**:
  Mereka mengubah model probabilistik prior secara artifisial. Ketika lokasi titik induksi $Z$ dioptimasi dengan gradient descent, seringkali terjadi **overfitting** (titik $Z$ menumpuk di area tertentu atau noise variance mengecil ke nol).

---

#### 3. Variational Sparse GP (Titsias 2009): Revolusi Inferensi Variasional
Michalis Titsias (pembimbing Andreas Damianou) pada tahun 2009 membuat terobosan revolusioner di paper *“Variational Learning of Inducing Variables in Sparse Gaussian Processes”*:
1. **Model prior asli tidak diubah sama sekali**: GP tetap memiliki prior $p(F, U) = p(F \mid U) p(U)$.
2. Menggunakan **Variational Inference** dengan mendefinisikan distribusi aproksimasi posterior:
   $$q(F, U) = p(F \mid U, X, Z) \, q(U)$$
   Di mana $p(F \mid U, X, Z) = \mathcal{N}(F \mid K_{NM}K_{MM}^{-1}U, \, K_{NN} - K_{NM}K_{MM}^{-1}K_{MN})$ diambil tepat sama dengan conditional prior aslinya!
3. Menurunkan batas bawah evidensi (**ELBO** - *Evidence Lower Bound*):
   $$\log p(Y \mid X) \ge \mathcal{L}_{SGPR} = \log \mathcal{N}\left(Y \,\Big|\, 0, \, Q_{NN} + \sigma_y^2 I_N\right) - \frac{1}{2\sigma_y^2} \operatorname{Tr}(K_{NN} - Q_{NN})$$

> 💡 **Mengapa Formula Titsias Ini Jenius?**
> - **Bebas Overfitting**: $Z$ dan $U$ sekarang adalah **parameter variasional**, bukan hyperparameter model. Jika $Z$ dioptimasi, kita hanya memperketat batas bawah variasional mendekati nilai fungsi objektif aslinya!
> - **Suku Penalti Jejak Matriks $\operatorname{Tr}(K_{NN} - Q_{NN})$**: Suku ini mengukur seberapa banyak informasi kovariansi yang hilang akibat menggunakan $M$ titik representatif. Jika $M$ titik $Z$ sudah mewakili seluruh data $X$, maka $Q_{NN} \to K_{NN}$ dan penalti jejak bernilai nol!
> - **Kompleksitas**: Turun drastis menjadi $\mathcal{O}(NM^2)$, hanya perlu membalik matriks kecil $K_{MM}$ ($M \times M$).

---

#### 4. Variational GP-LVM: Membuka Gerbang ke Ruang Laten & Deep GP
Sekarang masuk ke kontribusi inti disertasi Andreas Damianou (Bab 3 & 4):
* Pada GP biasa dan Variational Sparse GP, kita berasumsi input $X$ **selalu diketahui secara pasti**.
* Namun dalam banyak kasus nyata, input $X$ adalah **variabel tersembunyi / laten** (*Latent Variable Model* / GP-LVM, Neil Lawrence 2004):
  $$\text{Kita hanya mengamati data } Y, \text{ dan ingin memodelkan } Y \text{ dari representasi laten } X \text{ yang berdimensi lebih rendah } (Q \ll D).$$
* Model generatifnya:
  $$p(Y) = \int p(Y \mid X) p(X) \, dX$$
  Dengan prior Gaussian pada ruang laten: $p(X) = \prod_{n=1}^N \mathcal{N}(x_n \mid 0, I_Q)$.

> ❓ **Tantangan Berat**:
> Integral di atas **sangat tidak mungkin dihitung secara eksak (intractable)** karena variabel $X$ berada di dalam matriks kernel non-linear $K_{NN}(X, X)$ (misalnya di dalam fungsi eksponensial kernel RBF: $\exp(-\frac{1}{2}(x_i - x_j)^2)$)!

#### Solusi Damianou: Variational Marginalisation dengan Statistik $\Psi$
Damianou memperkenalkan distribusi variasional pada variabel laten $X$:
$$q(X) = \prod_{n=1}^N \mathcal{N}(x_n \mid \mu_n, S_n)$$
Dengan menggabungkan formulasi inducing points Titsias dan mengintegralkan $q(X)$ secara analitik, Damianou berhasil menghitung batas bawah variasional secara *closed-form* melalui 3 besaran ekspektasi kernel:
1. $\psi_0 = \operatorname{Tr}\left(\mathbb{E}_{q(X)}[K_{NN}]\right) = \sum_{n=1}^N \mathbb{E}_{q(x_n)}[k(x_n, x_n)]$
2. $\Psi_1 = \mathbb{E}_{q(X)}[K_{NM}] \in \mathbb{R}^{N \times M}$
3. $\Psi_2 = \mathbb{E}_{q(X)}[K_{MN} K_{NM}] \in \mathbb{R}^{M \times M}$

Hasilnya adalah **Variational GP-LVM**: kita dapat memasukkan input yang berupa distribusi probabilitas (tidak pasti / *uncertain*) ke dalam fungsi non-linear GP, dan menghasilkan batas bawah analitik tanpa perlu sampling!

## 💬 Sesi 2: Memahami SoR, DTC, dan FITC — Kerangka Kerja Sparse GP Klasik

### ❓ Pertanyaan Mahasiswa
> *"Apa yang anda maksud dengan FITC, DTC, dan SoR?"*

### 💡 Penjelasan & Derivasi Matematis
**SoR**, **DTC**, dan **FITC** adalah singkatan dari tiga metode aproksimasi **Sparse Gaussian Process klasik** yang dipopulerkan dan disatukan (*unified framework*) dalam paper penting karya **Joaquin Quiñonero-Candela & Carl Edward Rasmussen (2005)**, dan dibahas secara detail oleh Andreas Damianou pada **Bab 2 (Halaman 40–42)** tesisnya.

Ketiganya diciptakan untuk memangkas komputasi invers kovariansi dari $\mathcal{O}(N^3)$ menjadi $\mathcal{O}(NM^2)$ menggunakan $M$ buah titik induksi (*inducing points*) $Z = [z_1, \dots, z_M]^\top$ dengan nilai fungsi laten $U = f(Z)$.

---

### 🔤 Kepanjangan & Intuisi Utama

| Singkatan | Kepanjangan | Makna Intuisi Sederhana |
|---|---|---|
| **SoR** | *Subset of Regressors* | Menganggap fungsi laten $F$ **sepenuhnya deterministik** dari $U$, tanpa ada variansi residual sama sekali (seperti regresi basis linier). |
| **DTC** | *Deterministic Training Conditional* | Mengabaikan variansi residual pada data *training* (matriks kovariansi disederhanakan murni menjadi $Q_{NN}$), tetapi menghitung variansi pada data *test*. |
| **FITC** | *Fully Independent Training Conditional* | Mengasumsikan data observasi **saling bebas (independen)** jika $U$ diketahui. Suku diagonal kovariansi dipertahankan persis, kovariansi silang didekati via $Q_{NN}$. |

---

### 📐 Formulasi Matematis & Perbedaan Asumsinya

Mari kita lihat dari mana ketiga metode ini berasal secara matematika:

Pada GP eksak, jika kita menyisipkan $M$ titik induksi $U$, distribusi fungsi laten training $F$ terkondisi pada $U$ adalah:
$$p(F \mid U) = \mathcal{N}\left(F \,\Big|\, K_{NM} K_{MM}^{-1} U, \; \underbrace{K_{NN} - K_{NM} K_{MM}^{-1} K_{MN}}_{\widetilde{K}_{NN} \text{ (kovariansi residual)}}\,\right)$$
Di mana $Q_{NN} = K_{NM} K_{MM}^{-1} K_{MN}$ adalah aproksimasi berorde rendah (*low-rank Nyström approximation*) dari matriks kovariansi $K_{NN}$.

Ketiga metode klasik ini memodifikasi kovariansi residual $\widetilde{K}_{NN}$ dengan asumsi $\widetilde{Q}$ yang berbeda:

$$q(F \mid U) = \mathcal{N}(F \mid K_{NM} K_{MM}^{-1} U, \; \widetilde{Q})$$

#### 1. SoR (*Subset of Regressors*)
* **Asumsi Residual**: $\widetilde{Q} = 0$.
* Fungsi laten $f(x)$ diasumsikan linier terhadap kernel titik induksi:
  $$f(x_n) \approx k(x_n, Z) K_{MM}^{-1} U$$
* **Matriks Kovariansi Prior Efektif**:
  $$K_{NN}^{\text{SoR}} = Q_{NN} = K_{NM} K_{MM}^{-1} K_{MN}$$
* **Kelemahan Fatal**: Karena tidak ada ketidakpastian residual, variansi prediksi di luar titik induksi menjadi sangat sempit secara keliru (*grossly overconfident*).

#### 2. DTC (*Deterministic Training Conditional*) — Csató & Opper (2002), Seeger et al. (2003)
* **Asumsi Residual pada Training**: Sama seperti SoR, $\widetilde{Q} = 0$ untuk data latih:
  $$q_{\text{DTC}}(F \mid U, X) = \mathcal{N}(F \mid K_{NM} K_{MM}^{-1} U, \; 0)$$
* **Marginal Likelihood Training (Persamaan 2.22 tesis Damianou)**:
  $$q_{\text{DTC}}(Y \mid X) = \mathcal{N}\left(Y \,\Big|\, 0, \; Q_{NN} + \sigma_y^2 I_N\right)$$
* **Bedanya dengan SoR**: Pada saat memprediksi titik baru $x_*$ (testing), variansi residual $k(x_*, x_*) - q(x_*, x_*)$ tidak dibuang, sehingga ketidakpastian di area yang jauh dari data tetap membesar.
* **Kelemahan**: Menghilangkan variansi residual pada training membuat estimasi noise $\sigma_y^2$ seringkali bias dan rentan overfitting jika lokasi $Z$ digeser dengan gradient descent.

#### 3. FITC (*Fully Independent Training Conditional*) — Snelson & Ghahramani (2006)
* Awalnya diperkenalkan dengan nama **SPGP** (*Sparse Pseudo-input Gaussian Process*).
* **Ide Cerdas**: Mengapa membuang seluruh variansi residual seperti DTC? Yang membuat komputasi $\mathcal{O}(N^3)$ mahal hanyalah elemen **di luar diagonal** (*off-diagonal covariance*)!
* Maka FITC **mempertahankan elemen diagonal asli** dari kovariansi residual:
  $$\widetilde{Q} = \operatorname{diag}(K_{NN} - Q_{NN})$$
* **Matriks Kovariansi Prior Efektif**:
  $$K_{NN}^{\text{FITC}} = Q_{NN} + \operatorname{diag}(K_{NN} - Q_{NN})$$
* **Artinya**:
  - Variansi diri tiap data point ($k(x_n, x_n)$) dipulihkan secara tepat persis pada diagonal matriks.
  - Hanya korelasi antar-pasang data berbeda yang diproksikan lewat $M$ inducing points.
* **Kelebihan**: Variansi prediksi jauh lebih realistis dibanding DTC dan SoR.
* **Limitasi**: FITC tetap merupakan modifikasi prior artifisial, bukan bound variasional sejati pada GP asli. Jika $Z$ dioptimasi bersamaan dengan hyperparameter, model masih bisa mengalami *overfitting* lokal.

---

### 🏆 Mengapa Titsias (2009) & Damianou Memilih Pendekatan Berbeda?

Perhatikan hubungan magis antara **DTC** dan **Variational Sparse GP (Titsias)** pada Persamaan (2.33) di tesis Damianou:

$$\mathcal{L}_{\text{Variational}} = \underbrace{\log q_{\text{DTC}}(Y \mid X)}_{\text{Suku Likelihood DTC}} - \underbrace{\frac{1}{2\sigma_y^2} \operatorname{Tr}(K_{NN} - Q_{NN})}_{\text{Suku Penalti Jejak (Trace Regularizer)}}$$

* **Titsias menyadari**: DTC sebenarnya hampir benar, hanya saja DTC membuang variansi residual $\widetilde{K}_{NN} = K_{NN} - Q_{NN}$.
* Dengan Variational Inference, muncul suku koreksi otomatis yaitu jejak matriks $\operatorname{Tr}(K_{NN} - Q_{NN})$.
* Suku inilah yang **mengontrol agar model tidak overfitting**: jika $Z$ ditaruh di tempat yang sembarangan, nilai jejak $\operatorname{Tr}(K_{NN} - Q_{NN})$ akan membesar dan menurunkan nilai batas bawah (ELBO), sehingga optimizer dipaksa menyebarkan $Z$ secara optimal!

## 💬 Sesi 3: Silsilah Model — Apakah Metode Ini Modifikasi dari GP atau DGP?

### ❓ Pertanyaan Mahasiswa
> *"Dari metode-metode yang kamu sebutkan, apakah mereka adalah versi modifikasi dari GP ataupun DGP?"*

### 💡 Penjelasan & Derivasi Matematis
Jawaban singkat dan tegasnya:  
**Metode-metode tersebut (SoR, DTC, FITC, Variational Sparse GP, dan Variational GP-LVM) adalah modifikasi dari GP biasa (1 layer), BUKAN modifikasi dari DGP.**

Bahkan, hubungannya justru sebaliknya:  
👉 **DGP (Deep Gaussian Process) adalah arsitektur bertingkat yang DIBANGUN DENGAN MENGGABUNGKAN metode-metode GP tersebut sebagai batu bata penyusunnya (*building blocks*)!**

---

### 🌳 Pohon Silsilah / Genealogi: Dari GP Standar Menuju DGP

Untuk melihat posisi masing-masing metode secara gamblang, perhatikan diagram pohon evolusi berikut:

```
                          [Standard GP Regression]
                       (1 Layer, X -> Y, Eksak O(N^3))
                                     │
         ┌───────────────────────────┴───────────────────────────┐
         ▼                                                       ▼
[JALUR 1: Mengatasi Masalah Skalabilitas]         [JALUR 2: Mengatasi Sifat Input X]
Bagaimana agar hemat komputasi untuk data besar?    Bagaimana jika input X tidak berlabel/laten?
         │                                                       │
         ├─► SoR  (1985/2001) - Modifikasi Prior                 └─► GP-LVM (Lawrence 2004)
         ├─► DTC  (2002/2003) - Modifikasi Prior                       (Input X diestimasi via MAP point estimate,
         ├─► FITC (2005/2006) - Modifikasi Prior                        rentan overfitting)
         │                                                       │
         └─► Variational Sparse GP (Titsias 2009)                │
             (Batas bawah ELBO sejati, anti-overfitting)          │
                         │                                       │
                         └───────────────────┬───────────────────┘
                                             ▼
                             [Variational GP-LVM & Uncertain Inputs]
                                  (Titsias & Lawrence 2010; Damianou Bab 3 & 4)
                                  - Tetap 1 Layer GP
                                  - Input X diperlakukan sebagai distribusi probabilistik q(X)
                                  - Diintegralkan analitik menggunakan statistik Psi (Psi0, Psi1, Psi2)
                                             │
                                             ▼
                          =======================================
                           🚀 [DEEP GAUSSIAN PROCESS (DGP)] 🚀
                                (Damianou & Lawrence 2013; Tesis Bab 6)
                          =======================================
                           - Terdiri dari L-layer hierarkis:
                             X -> H_1 -> H_2 -> ... -> H_{L-1} -> Y
                           - Tiap layer 'l' adalah sebuah Variational GP-LVM!
```

---

### 🧩 Peran Masing-Masing Metode sebagai "Batu Bata" Pembentuk DGP

Mari kita telusuri bagaimana metode-metode 1-layer GP tersebut akhirnya melahirkan DGP:

1. **Mengapa DGP tidak bisa dibangun hanya dari GP Standar?**
   - Di GP standar, input $X$ wajib berupa nilai angka pasti (deterministik).
   - Di Deep GP, output dari Layer 1 ($H_1$) adalah variabel acak Gaussian yang memiliki ketidakpastian (mean $\mu$ dan varians $\sigma^2$).
   - Jika GP biasa dipaksa menerima input berdistribusi $H_1$, matriks kernel $K(H_1, H_1)$ akan menjadi variabel acak yang **terjebak di dalam operasi invers eksak $\mathcal{O}(N^3)$**, sehingga integralnya sama sekali tidak bisa diselesaikan (*intractable*).

2. **Peran Variational Sparse GP (Titsias 2009) pada DGP:**
   - Menyediakan konsep **inducing points $Z$**.
   - Inducing points ini bertindak sebagai perantara (*bottleneck*) berukuran kecil ($M \ll N$).
   - Akibatnya, ketergantungan antar-data berhasil diurai (*decoupled*), dan komputasi per layer menjadi hemat $\mathcal{O}(NM^2)$.

3. **Peran Variational GP-LVM & Statistik $\Psi$ (Damianou Bab 3 & 4) pada DGP:**
   - Menyediakan rumus analitik untuk melewatkan distribusi Gaussian $q(H_{l-1})$ menembus fungsi kernel non-linear layer ke-$l$.
   - Tiga rumus ajaib $\psi_0, \Psi_1, \Psi_2$ yang diturunkan Damianou membuat integrasi batas bawah ELBO di seluruh layer dapat dihitung secara tertutup (*closed-form*).

## 💬 Sesi 4: Bentuk Matriks Kovariansi K(H1, H1) & Contoh Operasi Angka Konkret

### ❓ Pertanyaan Mahasiswa
> *"Memangnya K(H1, H1) bentuknya seperti apa? tolong berikan contohnya dalam operasi matrix dengan angka"*

### 💡 Penjelasan & Derivasi Matematis
Mari kita bedah secara konkret dengan **angka nyata** apa itu $K(H_1, H_1)$, bagaimana wujud operasinya, dan **mengapa matriks ini menjadi sumber bencana matematika terbesar** jika kita mencoba menumpuk GP biasa secara naif.

---

### 1. Titik Pembanding: $K(X, X)$ pada GP Standar (Input Berupa Angka Pasti)

Misalkan kita memiliki $N = 3$ data observasi dengan nilai input $X$ yang **pasti / deterministik**:
$$x_1 = 1.0, \quad x_2 = 2.0, \quad x_3 = 3.0$$
Kita menggunakan kernel RBF (Gaussian) standar dengan panjang skala $\ell = 1.0$ dan varians sinyal $\sigma_f^2 = 1.0$:
$$k(x_i, x_j) = \exp\left( -\frac{(x_i - x_j)^2}{2} \right)$$

Kita hitung nilai setiap elemen matriks kovariansi $K(X, X)$:
* $k(x_1, x_1) = \exp\left( -\frac{(1 - 1)^2}{2} \right) = \exp(0) = 1.0000$
* $k(x_1, x_2) = \exp\left( -\frac{(1 - 2)^2}{2} \right) = \exp(-0.5) \approx 0.6065$
* $k(x_1, x_3) = \exp\left( -\frac{(1 - 3)^2}{2} \right) = \exp(-2.0) \approx 0.1353$

Maka matriks $K(X, X)$ berukuran $3 \times 3$ adalah **matriks angka biasa yang statis dan pasti**:
$$K(X, X) = \begin{bmatrix}
1.0000 & 0.6065 & 0.1353 \\
0.6065 & 1.0000 & 0.6065 \\
0.1353 & 0.6065 & 1.0000
\end{bmatrix}$$

Karena semua elemennya adalah angka desimal tetap, menghitung invers $[K(X, X) + \sigma_y^2 I]^{-1}$ dan determinannya sangat mudah dilakukan oleh komputer.

---

### 2. Apa yang Terjadi pada Deep GP di Layer 1? ($H_1$ Muncul)

Dalam Deep GP, input $X$ melewati GP Layer 1.  
Ingat bahwa **output dari sebuah GP bukanlah angka skalar tunggal**, melainkan **distribusi probabilitas Gaussian** yang memiliki rata-rata (mean $\mu$) dan ketidakpastian (varians $s^2$):
$$h_n^{(1)} \sim \mathcal{N}(\mu_n, s_n^2)$$

Misalkan untuk ketiga titik data di atas, output dari Layer 1 adalah:
* **Titik 1**: $h_1^{(1)} \sim \mathcal{N}(\mu_1 = 1.0, \; s_1^2 = 0.04)$ *(standar deviasi $\sigma = 0.2$)*
* **Titik 2**: $h_2^{(1)} \sim \mathcal{N}(\mu_2 = 2.0, \; s_2^2 = 0.09)$ *(standar deviasi $\sigma = 0.3$)*
* **Titik 3**: $h_3^{(1)} \sim \mathcal{N}(\mu_3 = 3.0, \; s_3^2 = 0.01)$ *(standar deviasi $\sigma = 0.1$)*

Sekarang, variabel $H_1 = [h_1^{(1)}, h_2^{(1)}, h_3^{(1)}]^\top$ ini dijadikan **INPUT** untuk Layer 2!

---

### 3. Wujud Sebenarnya dari $K(H_1, H_1)$

Matriks kovariansi di Layer 2 adalah:
$$K(H_1, H_1) = \begin{bmatrix}
k(h_1, h_1) & k(h_1, h_2) & k(h_1, h_3) \\
k(h_2, h_1) & k(h_2, h_2) & k(h_2, h_3) \\
k(h_3, h_1) & k(h_3, h_2) & k(h_3, h_3)
\end{bmatrix}$$

Mari kita zoom-in ke salah satu elemen di luar diagonal, misalnya **baris 1 kolom 2**:
$$[K(H_1, H_1)]_{1,2} = \exp\left( -\frac{(h_1^{(1)} - h_2^{(1)})^2}{2} \right)$$

> ⚠️ **PERHATIKAN KEJANGGALAN BESAR INI:**
> - $h_1$ dan $h_2$ **BUKAN ANGKA**, melainkan variabel acak Gaussian!
> - Selisih $(h_1 - h_2)$ adalah variabel acak.
> - Kuadrat $(h_1 - h_2)^2$ adalah variabel acak non-sentral Chi-Square.
> - $\exp(-\dots)$ adalah transformasi eksponensial non-linear dari variabel acak tersebut.
>
> **Artinya: Setiap elemen dari matriks $K(H_1, H_1)$ adalah sebuah VARIABEL ACAK / DISTRIBUSI PROBABILITAS, bukan angka pasti!**

#### Ilustrasi Angka Nyata Jika Kita Lakukan Sampling:
Karena $H_1$ acak, jika kita mengambil 2 sampel realisasi acak yang berbeda dari distribusi $H_1$, matriks $K(H_1, H_1)$ akan menghasilkan angka yang **selalu berubah-ubah**:

* **Sampel Acak Realisasi A**:
  Misalkan terambil nilai $h_1 = 1.0993, \; h_2 = 1.9585, \; h_3 = 3.0648$.
  $$K(H_1, H_1)^{(\text{sampel A})} = \begin{bmatrix}
  1.0000 & 0.6914 & 0.1449 \\
  0.6914 & 1.0000 & 0.5423 \\
  0.1449 & 0.5423 & 1.0000
  \end{bmatrix}$$

* **Sampel Acak Realisasi B**:
  Misalkan pada detik berikutnya terambil nilai $h_1 = 1.3046, \; h_2 = 1.9298, \; h_3 = 2.9766$.
  $$K(H_1, H_1)^{(\text{sampel B})} = \begin{bmatrix}
  1.0000 & 0.8225 & 0.2472 \\
  0.8225 & 1.0000 & 0.5781 \\
  0.2472 & 0.5781 & 1.0000
  \end{bmatrix}$$

---

### 4. Mengapa Ini "Bencana Matematika" pada GP Standar?

Pada GP standar, marginal likelihood dihitung dengan menginvers matriks kovariansi:
$$p(Y \mid H_1) = \mathcal{N}\left(Y \,\Big|\, 0, \, K(H_1, H_1) + \sigma_y^2 I_N\right) = \frac{1}{\sqrt{(2\pi)^N \det(K + \sigma_y^2 I)}} \exp\left(-\frac{1}{2} Y^\top [K(H_1, H_1) + \sigma_y^2 I]^{-1} Y\right)$$

Untuk membuang (marginalisasi) ketidaktahuan kita terhadap $H_1$, dalam kaidah Bayesian kita **wajib menghitung integral** terhadap distribusi $p(H_1)$:
$$p(Y) = \int p(Y \mid H_1) p(H_1) \, dH_1$$

Coba perhatikan posisi $H_1$:
1. $H_1$ berada di dalam operasi kuadrat $(h_i - h_j)^2$.
2. Operasi kuadrat itu berada di dalam fungsi eksponensial $\exp(\cdot)$.
3. Fungsi eksponensial itu menjadi elemen-elemen matriks $K(H_1, H_1)$.
4. Matriks itu kemudian **di-invers $[\dots]^{-1}$** dan dihitung **determinannya $\det(\dots)$**!
5. Dan seluruh operasi aljabar linier rumit ini harus diintegralkan terhadap distribusi Gaussian $p(H_1)$!

> 💥 **Kenyataan Matematis**:
> Di seluruh literatur kalkulus dan aljabar linier, **TIDAK PERNAH ADA rumus analitik (closed-form) untuk mengintegralkan invers matriks yang elemen-elemennya berupa eksponensial acak ($[K(H_1, H_1)]^{-1}$)**. Masalah ini disebut **intractable** (buntu secara analitik).

---

### 5. Bagaimana Damianou Menyelesaikannya Menggunakan Inducing Points $Z$?

Inilah trik brilian Andreas Damianou (Bab 3 & 6 Tesis):

1. **Damianou TIDAK PERNAH membalik matriks $K(H_1, H_1)$!**  
   Ia menyisipkan $M$ buah **Inducing Points $Z$** yang lokasinya dipilih berupa **angka pasti (deterministik)**!  
   Misal kita pilih $M = 2$ titik induksi pada Layer 2:
   $$z_1 = 1.5, \quad z_2 = 2.5$$

2. Karena $Z$ adalah angka pasti, maka matriks kovariansi $K(Z, Z) = K_{MM}$ adalah **matriks angka biasa**:
   $$K_{MM} = \begin{bmatrix}
   \exp(0) & \exp(-0.5) \\
   \exp(-0.5) & \exp(0)
   \end{bmatrix} = \begin{bmatrix}
   1.0000 & 0.6065 \\
   0.6065 & 1.0000
   \end{bmatrix}$$
   Matriks ini sangat mudah dibalik:
   $$K_{MM}^{-1} = \begin{bmatrix}
   1.582 & -0.960 \\
   -0.960 & 1.582
   \end{bmatrix}$$

3. Variabel acak $H_1$ sekarang **hanya berinteraksi dengan $Z$** melalui ekspektasi kernel:
   $$\Psi_1 = \mathbb{E}_{q(H_1)}[K(H_1, Z)], \quad \Psi_2 = \mathbb{E}_{q(H_1)}[K(Z, H_1) K(H_1, Z)]$$
   Karena integrasi ekspektasi antara fungsi eksponensial $\exp(-(h - z)^2)$ dan Gaussian $\mathcal{N}(h \mid \mu, s^2)$ adalah perkalian dua fungsi Gaussian, **HASIL INTEGRALNYA MEMILIKI RUMUS ANALITIK YANG INDAH**:
   $$[\Psi_1]_{n, m} = \frac{1}{\sqrt{1 + s_n^2 / \ell^2}} \exp\left( -\frac{(\mu_n - z_m)^2}{2(\ell^2 + s_n^2)} \right)$$

Perhatikan betapa cantiknya formula analitik di atas:
Ketidakpastian $s_n^2$ dari layer sebelumnya **secara elegan memperlebar panjang skala (lengthscale) kernel** menjadi $(\ell^2 + s_n^2)$ dan menurunkan puncaknya dengan faktor redaman $\frac{1}{\sqrt{1 + s_n^2 / \ell^2}}$! Semua masalah terselesaikan secara eksak tanpa perlu invers matriks acak.


````python
# Simulasi Komputasi Angka Konkret K(X, X) vs K(H1, H1) vs Statistik Psi Damianou
import numpy as np

# 1. GP Standar (Input Deterministik X)
x = np.array([1.0, 2.0, 3.0])
dist_sq_X = (x[:, None] - x[None, :]) ** 2
K_XX = np.exp(-0.5 * dist_sq_X)
print("=== 1. K(X, X) Deterministik (Angka Pasti) ===")
print(np.round(K_XX, 4))

# 2. Output Layer 1 (H1 adalah variabel acak Gaussian: mean & variance)
mu = np.array([1.0, 2.0, 3.0])
var = np.array([0.04, 0.09, 0.01])  # variansi ketidakpastian s_n^2

# 3. K(H1, H1) acak: dua kali sampling menghasilkan matriks angka berbeda
np.random.seed(42)
h1_sample_A = np.random.normal(mu, np.sqrt(var))
K_H1_sample_A = np.exp(-0.5 * (h1_sample_A[:, None] - h1_sample_A[None, :]) ** 2)

h1_sample_B = np.random.normal(mu, np.sqrt(var))
K_H1_sample_B = np.exp(-0.5 * (h1_sample_B[:, None] - h1_sample_B[None, :]) ** 2)

print("\n=== 2. Realisasi K(H1, H1) Sampel A (Berubah-ubah) ===")
print("Realisasi h1:", np.round(h1_sample_A, 4))
print(np.round(K_H1_sample_A, 4))

print("\n=== 3. Realisasi K(H1, H1) Sampel B (Berubah-ubah) ===")
print("Realisasi h1:", np.round(h1_sample_B, 4))
print(np.round(K_H1_sample_B, 4))

# 4. Solusi Damianou: Inducing Points Z deterministik + Statistik Psi_1 Analitik
Z = np.array([1.5, 2.5])  # M = 2 inducing points pasti
ell = 1.0

# Rumus Analitik Damianou (Tesis Persamaan B.2 / Hal 48)
# [Psi_1]_{n, m} = E_{q(h_n)} [ k(h_n, z_m) ]
Psi_1 = np.zeros((len(mu), len(Z)))
for n in range(len(mu)):
    for m in range(len(Z)):
        scale = 1.0 / np.sqrt(1.0 + var[n] / (ell**2))
        exponent = -0.5 * ((mu[n] - Z[m])**2) / (ell**2 + var[n])
        Psi_1[n, m] = scale * np.exp(exponent)

print("\n=== 4. Solusi Damianou: Matriks Statistik Psi_1 (Analitik & Deterministik) ===")
print("Ukuran Psi_1 (N x M):", Psi_1.shape)
print(np.round(Psi_1, 4))
````

## 💬 Sesi 5: Apakah Inducing Points Satu-Satunya Metode untuk Menyelesaikan DGP? (Eksplorasi Metode Alternatif)

### ❓ Pertanyaan Mahasiswa
> *"Apakah inducing point adalah satu-satunya metode? Jika ada metode lain, apa itu?"*

### 💡 Penjelasan & Derivasi Matematis
Jawaban singkat dan tegasnya:  
**TIDAK, Inducing Points BUKAN satu-satunya metode.**

Meskipun *Inducing Points* (seperti yang digunakan Andreas Damianou dalam tesisnya) adalah pendekatan yang **paling dominan, paling sukses secara historis, dan menjadi standar baku**, komunitas riset *Machine Learning* dan komputasi statistik telah mengembangkan beberapa **paradigma alternatif** yang sangat berbeda secara filosofis maupun matematis.

---

### 🎯 Mengapa Masalah Tersebut Butuh Solusi? (Mengingat Akar Masalahnya)
Ingat kembali dari Sesi 4: Masalah mendasar pada layer tersembunyi DGP adalah ketika input ke layer ke-$l$ adalah variabel acak $H \sim \mathcal{N}(\mu, S)$, matriks kovariansi $K(H, H)$ menjadi matriks acak.
Untuk menghitung *marginal likelihood*, kita terbentur pada integral tak terhitung (*intractable*):
$$p(Y) = \int \mathcal{N}\left(Y \mid 0, K(H, H) + \sigma^2 I\right) p(H) \, dH$$
Operasi invers $[K(H, H) + \sigma^2 I]^{-1}$ dan determinan dari matriks eksponensial acak ini **tidak memiliki solusi analitik (closed-form)** di dunia kalkulus.

Untuk menaklukkan kebuntuan ini, para peneliti membaginya ke dalam **4 sudut pandang strategi**:
1. **Strategi 1 (Aproksimasi Fitur)**: Mengubah representasi kernel tak hingga menjadi fitur linier berhingga $\implies$ **Random Fourier Features (RFF)**.
2. **Strategi 2 (Sampling Numerik Murni)**: Mengambil sampel angka konkret sehingga matriksnya statis $\implies$ **Markov Chain Monte Carlo (MCMC / SGHMC)**.
3. **Strategi 3 (Pencocokan Momen)**: Mengintegralkan momen analitik lalu memproyeksikan kembali ke Gaussian $\implies$ **Moment Matching (Girard/PILCO)**.
4. **Strategi 4 (Variational Reformulation)**: Merekayasa variasi stochastic tanpa integrasi analitik $\implies$ **Doubly Stochastic VI (Salimbeni & Deisenroth, 2017)**.

---

### 🔍 5 Metode Alternatif Utama dalam Literatur

#### 1. Random Fourier Features (RFF) / Spectral Kernel Expansion (Cutajar et al., 2017)
* **Status Inducing Points**: ❌ **SAMA SEKALI TIDAK MENGGUNAKAN INDUCING POINTS!**
* **Konsep Inti**:
  Berdasarkan **Teorema Bochner (1959)**, fungsi kernel stasioner seperti RBF dapat dinyatakan sebagai transformasi Fourier dari distribusi spektral frekuensi:
  $$k(x, x') = \int p(\omega) e^{i \omega^\top (x - x')} d\omega \approx \frac{1}{D_{\text{rff}}} \sum_{d=1}^{D_{\text{rff}}} \phi_{\omega_d}(x)^\top \phi_{\omega_d}(x')$$
  Di mana $\phi(x) = \sqrt{\frac{2}{D_{\text{rff}}}} \cos(W x + b)$ dengan bobot acak $W \sim \mathcal{N}(0, \ell^{-2} I)$ dan bias $b \sim \text{Uniform}(0, 2\pi)$.
* **Bagaimana Mengatasi Bencana $K(H, H)$?**
  Alih-alih membangun matriks kovariansi $N \times N$, data dipetakan ke matriks fitur $\Phi(H) \in \mathbb{R}^{N \times D_{\text{rff}}}$.
  Fungsi laten berubah menjadi regresi linier berbobot Bayesian:
  $$f(H) \approx \Phi(H) \mathbf{w}, \quad \mathbf{w} \sim \mathcal{N}(0, I)$$
  **Hasilnya**: Deep GP berubah wujud menjadi **Deep Bayesian Neural Network (BNN)** dengan fungsi aktivasi kosinus periodik acak!
* **Operasi Matriks**:
  Kita tidak pernah membalik matriks $N \times N$. Yang dibalik hanyalah matriks berdimensi kecil $D_{\text{rff}} \times D_{\text{rff}}$ (misalnya 50 atau 100 fitur acak), berapa pun jumlah data $N$-nya!
* **Kelebihan**: Sangat cepat, komputasinya seperti Neural Network biasa, sangat mudah di-GPU-kan.
* **Kekurangan**: Sifat non-parametrik GP yang kaya sedikit terdegradasi menjadi parametrik berdimensi $D_{\text{rff}}$. Membutuhkan banyak fitur acak untuk memodelkan batas keputusan yang sangat tajam.

---

#### 2. Markov Chain Monte Carlo (MCMC) & Stochastic Gradient HMC (Havasi et al., 2018)
* **Status Inducing Points**: ❌ **Bisa murni tanpa Inducing Points** (atau opsional dipadukan untuk skala besar).
* **Konsep Inti**:
  Metode Bayesian sejati tanpa aproksimasi variational (tidak butuh mengasumsikan distribusi pendekatan $q(H)$). Menggunakan algoritma **Hamiltonian Monte Carlo (HMC)** atau **Stochastic Gradient HMC (SGHMC)** untuk menarik sampel langsung dari posterior sejati $p(H_1, H_2, \dots \mid Y)$.
* **Bagaimana Mengatasi Bencana $K(H, H)$?**
  Pada setiap langkah rantai sampling (*Markov chain step*), variabel laten $H_1$ **ditarik sebagai satu set nilai angka riil konkret** (misalnya $h_1^{(1)} = 1.092, h_2^{(1)} = 2.115$).
  Karena pada detik itu $H_1$ berwujud angka desimal statis, **matriks $K(H_1, H_1)$ otomatis menjadi matriks angka biasa** (persis seperti GP regresi standar)! Tidak ada lagi integral terhadap variabel acak di dalam eksponensial.
* **Kelebihan**: Asimtotik eksak (*gold standard* akurasi), tidak mengalami bias optimasi ELBO, dapat menangkap bentuk posterior multi-modal (memiliki banyak puncak probabilitas).
* **Kekurangan**: Komputasi luar biasa lambat, membutuhkan ratusan iterasi *burn-in*, dan sulit dipantau konvergensinya pada arsitektur layer yang dalam.

#### 3. Moment Matching / Gaussian Analytical Projection (Girard et al., 2003; PILCO - Deisenroth 2011)
* **Status Inducing Points**: ❌ **Tanpa Inducing Points pada formulasi aslinya**.
* **Konsep Inti**:
  Ini adalah teknik klasik yang sangat terkenal di bidang robotika dan *reinforcement learning*. Ketika input Gaussian $x \sim \mathcal{N}(\mu, \Sigma)$ dilewatkan ke GP dengan kernel RBF, distribusi output $f(x)$ sebenarnya bukan lagi Gaussian.
  Namun, kita bisa **menghitung rata-rata eksak (momen ke-1)** dan **kovariansi eksak (momen ke-2)** dari output tersebut secara analitik menggunakan konvolusi Gaussian:
  $$\mu_f = \mathbb{E}_{x \sim \mathcal{N}}[f(x)], \quad \sigma_f^2 = \operatorname{Var}_{x \sim \mathcal{N}}[f(x)]$$
* **Bagaimana Mengatasi Bencana $K(H, H)$?**
  Setelah momen $\mu_f$ dan $\sigma_f^2$ dihitung, distribusinya **dipaksa (*projected*) kembali menjadi distribusi Gaussian $\mathcal{N}(\mu_f, \sigma_f^2)$**, lalu diteruskan sebagai input deterministik-statistik ke layer berikutnya secara rekursif.
* **Kelebihan**: Sepenuhnya analitik dan deterministik, tidak memerlukan sampling angka acak Monte Carlo.
* **Kekurangan**: Kompleksitas tetap $\mathcal{O}(N^3)$ per layer (tidak menyelesaikan masalah data besar tanpa teknik tambahan), dan membuang informasi ketidakpastian momen tingkat tinggi (*skewness* dan *kurtosis*).

#### 4. Expectation Propagation (EP / Power-EP) (Hernández-Lobato et al., 2016)
* **Status Inducing Points**: Menggunakan varian *Stochastic Expectation Propagation* (SEP).
* **Konsep Inti**:
  Alih-alih memaksimalkan batas bawah variasional (ELBO) seperti Titsias dan Damianou, metode ini menggunakan kerangka kerja **Expectation Propagation (EP)**. EP memecah faktor *likelihood* yang tidak terhitung menjadi suku-suku aproksimasi lokal dari keluarga eksponensial (Gaussian), lalu mencocokkan momen (*moment matching*) pada *cavity distributions* secara bergantian.
* **Kelebihan**: Secara empiris menghasilkan estimasi variansi prediktif yang lebih realistis dibanding Variational Inference yang cenderung *under-estimate* (terlalu sempit).
* **Kekurangan**: Tidak ada jaminan konvergensi matematis (kadang bisa mengalami osilasi/divergensi), dan penurunannya jauh lebih rumit daripada ELBO.

---

#### 5. Doubly Stochastic Variational Inference (DSVI) — Salimbeni & Deisenroth (NeurIPS 2017)
> 🌟 **PENTING UNTUK DIKETAHUI MAHASISWA TA**: Ini adalah metode DGP paling modern dan paling banyak dipakai saat ini di industri & riset (menjadi fondasi engine library **GPyTorch** dan **GPflow**).

* **Status Inducing Points**: ✅ **Tetap menggunakan Inducing Points $Z$, TETAPI membuang Statistik $\Psi$ Damianou!**
* **Mengapa Metode Ini Muncul?**
  Pada tesis Andreas Damianou (2015), inducing points dikawinkan dengan **Statistik $\Psi$ analitik**. Kelemahannya:
  1. Hanya bisa berjalan pada kernel RBF dan Linear (karena kernel lain seperti Matérn atau Periodik tidak memiliki rumus analitik untuk $\Psi_1$ dan $\Psi_2$).
  2. Semua data $N$ saling terikat dalam satu matriks besar (sulit melakukan *mini-batch* SGD seperti pada Neural Network).
* **Terobosan Salimbeni & Deisenroth**:
  Mereka membuktikan bahwa jika kita memanfaatkan **Reparameterization Trick** (seperti pada VAE / *Variational Autoencoder*):
  $$h_n^{(l)} = \mu_n^{(l)} + \sqrt{v_n^{(l)}} \odot \epsilon, \quad \epsilon \sim \mathcal{N}(0, I)$$
  maka inferensi variasional dapat dilakukan **secara independen per titik data (decoupled across data points)**!
  - Kita tidak perlu lagi rumus analitik $\Psi$ yang rumit.
  - Kita bisa menggunakan **SEMUA JENIS KERNEL** (Matérn-3/2, Matérn-5/2, Cauchy, dll.).
  - Kita bisa melatih DGP dengan jutaan data menggunakan **Mini-batch Stochastic Gradient Descent di GPU**!
